Data cleaning
- Duplicate orders identify karein aur investigate karke duplicates remove karein.
- Missing Quantity, Unit_Price_PKR, Discount_Rate aur City ko handle karein.
- City names ko standardize karein, jaise lahore aur Karachi .
- Order_Date ko proper datetime format mein convert karein.

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
```

```python
df['Unit_Price_PKR'] = df['Unit_Price_PKR'].fillna(df['Unit_Price_PKR'].median())
```

```python
df['Discount_Rate'] = df['Discount_Rate'].fillna(df['Discount_Rate'].median())
```

```python
df['City'] = df['City'].fillna('unknown')
```

```python
df['City'] = df['City'].str.strip().str.title()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
```

```python
df.columns
```

Question 3: Sales Calculation
- Gross_Sales column banayein.
- Discount_Amount column banayein.
- Net_Sales calculate karein.
- Completed orders ki total net sales calculate karein.

```python
df['Gross_Sales'] = df['Quantity'] * df['Unit_Price_PKR']
```

```python
df['Discount_Amount'] = (df['Discount_Rate'] * df['Gross_Sales'] )/100
```

```python
df['Net_Sales'] = df['Gross_Sales'] - df['Discount_Amount']
```

```python
df[df['Order_Status'] == 'Completed']['Net_Sales'].sum()
```

Question 7: Monthly Sales Trend
- Har month ki sales calculate karein.
- Sabse zyada sales wala month find karein.
- Monthly revenue aur order count compare karein.

```python
df['Year_Month'] = df['Order_Date'].dt.to_period('M')
```

```python
df.groupby('Year_Month')['Net_Sales'].sum().idxmax()
```

```python
revenue = df.groupby('Year_Month')['Net_Sales'].sum()
orders = df.groupby('Year_Month')['Order_ID'].count()

comparison = pd.concat (
    [revenue, orders],
    axis=1,
    keys=['Monthly_Revenue', 'Order_Count']
)
comparison
```
